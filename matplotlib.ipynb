```python
import matplotlib.pyplot as plt
import numpy as np
products = np.array(["laptop","mobile","tablet","headphones","watch"])
sales = np.array([120,180,90,150,110])
profit = np.array([30,50,20,40,25])
print(products)
print(sales)
print(profit)


```

```python
#line chart - show trend over a time
plt.plot(products,sales,marker = "o")
plt.title("product sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
#multiple line chart -used to compare 2 or more trends
plt.plot(products,sales,marker = "o",label = "sales")
plt.plot(products,profit,marker = "o",label = "profit")
plt.title("product sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.legend()
plt.show()
```

```python
#bar chart--used cmpr values -- one-dime,one - meas
plt.bar(products,sales)
plt.title("product sales")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()

print("------------------------------------------------------------------------------------------------------")

#bar chart--used cmpr values -- one-dime,one - meas
plt.bar(sales,products)
plt.title("product sales")
plt.xlabel("sales")
plt.ylabel("products")
plt.show()
```

```python
#grouped bar chart(clustured)
x = np.arange(len(products))
width = 0.35
plt.bar(x - width/2,sales,width,label = "sales")
plt.bar(x + width/2,profit,width,label = "profit")
plt.xticks(x, products)
plt.title("sales vs profit")
plt.xlabel('products')
plt.ylabel("amount")
plt.legend()
plt.show()

print("------------------------------------------------------------------------------------------------")

#grouped bar chart
x = np.arange(len(products))
width = 0.35
plt.bar(x - 2/6,sales,width,label = "sales")
plt.bar(x + 2/6,profit,width,label = "profit")
plt.xticks(x, products)
plt.title("sales vs profit")
plt.xlabel('products')
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python
#stacked bar chart
x = np.arange(len(products))
width = 0.35
plt.bar(x - width/2,sales,width,label = "sales")
plt.bar(x - width/2,profit,width,label = "profit")
plt.xticks(x, products)
plt.title("sales vs profit")
plt.xlabel('products')
plt.ylabel("amount")
plt.legend()
plt.show()

print("---------------------------------------------------------------------------------------------------------------------------")

x = np.arange(len(products))
width = 0.35
plt.bar(x + width/2,sales,width,label = "sales")
plt.bar(x + width/2,profit,width,label = "profit")
plt.xticks(x, products)
plt.title("sales vs profit")
plt.xlabel('products')
plt.ylabel("amount")
plt.legend()
plt.show()
```

```python
#piechart-- evry part is knw as pie --360 deg and overall 100 each represnet an angle angle is 360 deg
plt.pie(
    sales,
    labels = products,  #in label only dim
    autopct= "%1.1f%%"  #automatic percentage -- evry pie represnt perc -- overall 100 -1
)
plt.title("product sales distribution")
plt.show()
```

```python
#donut chart--each part represnt knw as wedges--centre lo text rayachu by using plt.text---
plt.pie(
    sales,
    labels = products,  #in label only dim
    autopct= "%1.1f%%",  #automatic percentage -- evry pie represnt perc -- overall 100 -1
    wedgeprops={"width" :0.4}
)
plt.title("product sales distribution")
plt.show()
```

```python
#H.w
import matplotlib.pyplot as plt
products = ["Laptop", "Mobile", "Tablet", "Headphones", "Smartwatch"]
sales = [50, 80, 40, 30, 60]
departments = ["HR", "IT", "Sales", "Finance", "Marketing"]
employees = [10, 25, 20, 12, 15]
categories = ["Electronics", "Clothing", "Furniture", "Grocery"]
category_sales = [150, 100, 70, 50]
age_groups = ["18-25", "26-35", "36-45", "46-55", "56+"]
customers = [40, 60, 50, 30, 20]
print(products)
print(sales)
print(departments)
print(employees)
print(categories)
print(category_sales)
print(age_groups)
print(customers)
```

```python
#compare product sales -- Bar
plt.bar(products, sales)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales")
plt.show()
```

```python
#Product with highest and lowest sales
print("Highest sales:", products[sales.index(max(sales))])
print("Lowest sales:", products[sales.index(min(sales))])
plt.bar(products, sales)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales")
plt.show()
```

```python
#Compare employees in departments
plt.bar(departments, employees)
plt.xlabel("Departments")
plt.ylabel("Employees")
plt.title("Employees by Department")
plt.show()
```

```python
#Department with highest employees
print("Highest employees:", departments[employees.index(max(employees))])
plt.bar(departments, employees)
plt.xlabel("Departments")
plt.ylabel("Employees")
plt.title("Employees by Department")
plt.show()
```

```python
#how total sales distribution among categories
plt.pie(category_sales, labels=categories, autopct="%1.1f%%")
plt.title("Category Sales Distribution")
plt.legend()
plt.show()
```

```python
#Category contributing most to sales
print("Highest category:", categories[category_sales.index(max(category_sales))])
plt.pie(category_sales, labels=categories, autopct="%1.1f%%")
plt.title("Category Sales Distribution")
plt.legend()
plt.show()
```

```python
#Customer distribution across age groups
plt.bar(age_groups, customers)
plt.xlabel("Age Groups")
plt.ylabel("Customers")
plt.title("Customer Distribution")
plt.show()
```

```python
#Age group with highest customers
print("Highest customers:", age_groups[customers.index(max(customers))])
plt.bar(age_groups, customers)
plt.xlabel("Age Groups")
plt.ylabel("Customers")
plt.title("Customer Distribution")
plt.show()
```

```python
#sep-10
import pandas as pd
data = {

"student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
 "Math": [65, 70, 75, 80, 85, 60, 90, 72, 78, 88],
 "Python": [70, 75, 80, 85, 90, 65, 95, 78, 82, 92],
 "SQL": [60, 68, 72, 78, 82, 55, 88, 70, 75, 85]

}
df = pd.DataFrame(data)
print(df)
```

```python
#Violin Plot--Used to understand the distribution and density of data.
import matplotlib.pyplot as plt
plt.violinplot(df["Math"])
plt.title("Distribution of Math Marks")
plt.ylabel("Marks")
plt.show()
```

```python
#Stem Plot-- Useful for displaying discrete numerical values.
plt.stem(df["student"], df["Math"])
plt.title("Student Math Marks")
plt.xlabel("student")
plt.ylabel("marks")
plt.show()
```

```python
#contour plot
import numpy as np
import matplotlib.pyplot as plt
x = [1,2,3]
y = [10,20]
X,Y = np.meshgrid(x,y)
Z = X ** 2 + Y ** 2
plt.contour(X,Y,Z)
plt.title("contour plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#heat map--show diff clrs of visulization--high value - dark clr, low - lighter clr
#used methos --imshow(),colourbar - rept clr
marks = df[["Math","Python","SQL"]].values
plt.imshow(marks,cmap = "Reds") #to change clrs
plt.title("Student Marks HeatMap")
plt.xlabel("subjects")
plt.ylabel("students")
plt.colorbar()
```

```python
#3d line chart
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [100,150,200,250,300]
z = [20,30,40,50,60]
fig = plt.figure()
ax=fig.add_subplot(111, projection="3d")
ax.plot(x,y,z)
ax.set_xlabel("month")
ax.set_ylabel("sales")
ax.set_zlabel("profit")
ax.set_title("3D sales and profit")
plt.show()
```

```python
#3d plotting-- bar chart
import matplotlib.pyplot as plt
x= [1,2,3,4]
y = [1,2,3,4]
z = [0,0,0,0]
dx = [0.5,0.5,0.5,0.5]  #width
dy = [0.5,0.5,0.5,0.5] #depth
dz = [10,15,20,10]  #height
fig = plt.figure()  #to mention 3d fig
ax = fig.add_subplot(111,projection = "3d")
ax.bar3d(x,y,z,dx,dy,dz)  #postional agrumnts
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()

print("--------------------------------------------------------------------")

import matplotlib.pyplot as plt
x= [1,2,3,4]
y = [1,2,3,4]
z = [0,0,0,0]
#dx = [0.5,0.5,0.5,0.5]  #width
#dy = [0.5,0.5,0.5,0.5] #depth
#dz = [10,15,20,10]
fig = plt.figure()
ax = fig.add_subplot(111,projection = "3d")
ax.bar3d(x,y,z,0.5,0.5,[10,20,30,40]) #showing width,depth,height
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()
```
