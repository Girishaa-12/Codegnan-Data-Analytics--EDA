```python
#horizontal --side by side, two arr should hv same arr
arr1 = np.array([[10,20,30],
                [40,50,60]])
arr2 = np.array([[30,50,40],
                 [90,70,10]])
print(np.concatenate((arr1,arr2),axis = 1)) #col
```

```python
#vertical -- row wise
arr1 = np.array([[10,20],
                [40,50]])
arr2 = np.array([[30,50],
                 [90,70]])
print(np.concatenate((arr1,arr2),axis = 0)) #row
```

```python
#np.stack()--dim no change,
#1d
arr1 = np.array([10,20,30])
arr2 = np.array([50,40,60])
res = np.concatenate((arr1,arr2),axis = 0)
print(res)
print(res.ndim)

print("----------------------------------------------------")

#stack - new dim is created
#1d
arr1 = np.array([10,20,30])
arr2 = np.array([50,40,60])
res = np.stack((arr1,arr2),axis = 0)
print(res)
print(res.ndim)

print("--------------------------------------------")

#2D
arr = np.arange(10,20).reshape(2,5)
arr3 = np.arange(1,11).reshape(2,5)
print(arr)
print(arr3)
arr4=np.stack((arr,arr3))
print(arr4.ndim)

print("--------------------------------------------------")

#3d

arr = np.arange(10,20).reshape(2,5,1)
arr3 = np.arange(1,11).reshape(2,5,1)
print(arr)
print(arr3)
arr4=np.stack((arr,arr3))
print(arr4.ndim)

```

```python
#v-stack() --
#1d
arr1 = np.array([10,20,30])
arr2 = np.array([50,40,60])
res = np.vstack((arr1,arr2))
print(res)
print(res.ndim)


print("--------------------------------------------")

arr1 = np.array([10,20,30])
arr2 = np.array([50,40,60])
arr3 = np.array([34,56,78])
res = np.vstack((arr1,arr2,arr3))
print(res)
print(res.ndim)

print("-----------------------------------------------")

#2D
arr = np.arange(10,20).reshape(2,5)
arr3 = np.arange(1,11).reshape(2,5)
print(arr)
print(arr3)
arr4=np.vstack((arr,arr3))
print(arr4.ndim)


print("--------------------------------------------------")

#3d

arr = np.arange(10,20).reshape(2,5,1)
arr3 = np.arange(1,11).reshape(2,5,1)
print(arr)
print(arr3)
arr4=np.vstack((arr,arr3))
print(arr4.ndim)



```

```python
#H-stack() -- horizontal - side by side
#1d
arr1 = np.array([10,20,30])
arr2 = np.array([50,40,60])
res = np.hstack((arr1,arr2))
print(res)
print(res.ndim)

print("----------------------------------------")

arr1 = np.array([10,20,30])
arr2 = np.array([50,40,60])
arr3 = np.array([34,56,78])
res = np.hstack((arr1,arr2,arr3))
print(res)
print(res.ndim)


print("--------------------------------------------")

#2D
arr = np.arange(10,20).reshape(2,5)
arr3 = np.arange(1,11).reshape(2,5)
print(arr)
print(arr3)
arr4=np.hstack((arr,arr3))
print(arr4.ndim)


print("--------------------------------------------------")

#3d

arr = np.arange(10,20).reshape(2,5,1)
arr3 = np.arange(1,11).reshape(2,5,1)
print(arr)
print(arr3)
arr4=np.hstack((arr,arr3))
print(arr4.ndim)
```
