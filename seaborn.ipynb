```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
#load data
df = sns.load_dataset("titanic") #built in dataset
print(df)
```

```python
print(df.head(11))
```

```python
print(df.tail(11))
```

```python
print(df.info)
```

```python
print(df.describe)
```

```python
#line  --hue used for categorical values
sns.lineplot(data = df,x="pclass",y="fare",hue="sex",marker="*")
```

```python
#scatter
sns.scatterplot(data = df,x="pclass",y="fare")
```

```python
#bar
sns.barplot(data = df,x="pclass",y="fare",hue="sex")
```

```python
#heatmap  -- basic version -- single categriocal
corr = df[["age","fare","sibsp","parch"]].corr()
sns.heatmap(
    corr,
    annot=True
)
plt.show()
```

```python
#pivot table + heatmap -- related to pandas -- intermediate version
pivot = pd.pivot_table(
    df,
    values = "survived",
    index = "sex",
    columns = "class",
    aggfunc = "mean"
)
print(pivot)

sns.heatmap(
    pivot,
    annot = True,
    fmt = ".2f"
)
```

```python
#multi categ heat map
pivot = pd.pivot_table(
    df,
    values = "survived",
    index = ["sex","class"],
    aggfunc = "mean"
)
print(pivot)

sns.heatmap(
    pivot,
    annot = True,
    fmt=".2f"
)
```

```python
#countplot -- we can use from derive data --one dimen
sns.countplot(
    data = df,
    x="class"
)
plt.show()
```

```python
#regression plot
sns.regplot(
    data = df,
    x = "age",
    y = "fare"
)
plt.show()
```

```python
#violin plot
sns.violinplot(
    data = df,
    x = "class",
    y = "age",
    hue = "sex"
)
plt.title("Age Distributation By Class and Gender")
plt.show()
```

```python
#density plot
sns.kdeplot(
    data = df,
    x = "age",
    hue = "sex",
    fill = True
)
plt.title("Age Density by Gender")
plt.show()
```

```python
#pair plot -- 2 numerical data - multiple data we can see --count,hist,scatter --- adv version of scatters knw as pair plot
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.pairplot(
    df[["age","fare","sibsp"]]
)
plt.show()
```

```python
#factegrid
g = sns.FacetGrid(
    df,
   col = "sex",
)
g.map(
    sns.histplot,
    "age"
)
plt.show()

print("----------------------------------------------------------")

#factegrid
g = sns.FacetGrid(
    df,
   col = "sex",
)
g.map(
    sns.barplot,
    "age"
)
plt.show()

print("----------------------------------------------------------------------")

#factegrid
g = sns.FacetGrid(
    df,
   col = "sex"
)
g.map(
    sns.scatterplot,
    "age","fare"
)
plt.show()


```

```python
#joint plot--relation + distibuation = joint (scatter nd histo)
sns.jointplot(
    data = df,
    x = "age",
    y = "fare"
)
plt.show()
```
